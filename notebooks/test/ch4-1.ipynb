广播机制。两个形状不同的张量可以通过广播机制实现按元素操作

```python
a = torch.arange(3).reshape((3, 1))
b = torch.arange(2).reshape((1, 2))
a, b
```

a是一个三行一列的张量，b是一个一行二列的张量。维数相同，形状不同。仍然可以按元素操作。a将自身复制，成为一个三行二列的张量，b将自身复制，成为一个三行二列的张量，然后按元素操作

```python
a + b
```

```python
Y = torch.arange(12).reshape((3, 4))
Y
```

一些操作可能会导致为新结果分配新内存。用+=或者[:]进行赋值，可以保持内存地址不改变。

```python
before = id(Y)
Y = Y + X
id(Y) == before
```

```python
before = id(Y)
Y[:] = Y + X
id(Y) == before
```

```python
before = id(Y)
Y += X
id(Y) == before
```

```python
Z = torch.zeros_like(Y)
print('id(Z):', id(Z))
Z[:] = Y + X
print('id(Z):', id(Z))
```
